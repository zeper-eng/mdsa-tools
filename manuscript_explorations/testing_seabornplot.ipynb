```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns

# Load the data
RMSD_df = pd.read_csv("/Users/luis/Desktop/workspacetwo/manuscript_explorations/RMSD_dataframe.csv")

# Optional: split by system
GCU_system = RMSD_df[RMSD_df["system"] == 0]
CGU_system = RMSD_df[RMSD_df["system"] == 1]
```

```python

# Function to make the faceted plot
def plot_system(data, outname):
    g = sns.FacetGrid(
        data,
        col="replicate",
        col_wrap=10,
        sharey=True,
        sharex=True,
        height=2.0   # instead of 3
    )
    g.map_dataframe(
        sns.lineplot,
        x="window",
        y="RMSD",
        hue="cluster",
        palette="tab10"
    )
    g.add_legend()
    g.set_axis_labels("Window", "RMSD")
    plt.subplots_adjust(top=0.9)
    g.fig.suptitle("RMSD by replicate")
    plt.savefig(outname, dpi=800, bbox_inches="tight")
    plt.close()
```

```python
# Make both plots
plot_system(GCU_system, "GCU_RMSD_facetplot.png")
plot_system(CGU_system, "CGU_RMSD_facetplot.png")
```
